```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df["Month"] = df["Date"].dt.month_name()

df["Day"] = df["Date"].dt.day_name()
```

EXPLORATORY DATA ANALYSIS

```python
total_revenue = df["Sales"].sum()

print("Total Revenue:", total_revenue)
```

```python
average_sales = df["Sales"].mean()

print("Average Transaction Value:", average_sales)
```

```python
product_sales = df.groupby("Product line")["Sales"].sum()

print(product_sales.sort_values(ascending=False))
```

```python
branch_sales = df.groupby("Branch")["Sales"].sum()

print(branch_sales.sort_values(ascending=False))
```

```python
payment_count = df["Payment"].value_counts()

print(payment_count)
```

```python
customer_sales = df.groupby("Customer type")["Sales"].mean()

print(customer_sales)
```

```python
monthly_sales = df.groupby("Month")["Sales"].sum()

print(monthly_sales)
```
