What's happening?
There are two main components:
1. Preprocessor: Converts categorical variables into numerical values using One-Hot Encoding.
2. Linear Regression: Learns relationships between the input features and the target variable (Sales).
The .fit() method is where the model learns from the training data.

Evaluating the Linear Regression Model
Now we need to determine how well the model performs across the entire testing dataset, not just the first 10 transactions.

We'll use three evaluation metrics.
A. Mean Absolute Error (MAE)
MAE measures the average absolute difference between actual and predicted values.

For example, an MAE of 30 means that, on average, our predictions are approximately 30 currency units away from the actual sales.
B. Mean Squared Error (MSE)
MSE calculates the average squared prediction error.

It penalises larger errors more heavily than smaller ones.
C. R² Score
R² measures how much variation in the target variable is explained by the model.

Generally:
- 1.0 = perfect fit.
- 0 = no improvement over predicting the mean.
- Negative = worse than predicting the mean.

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

r2 = r2_score(y_test, y_pred)

print("Linear Regression Results")
print("-------------------------")
print("MAE:", mae)
print("MSE:", mse)
print("R2 Score:", r2)
```

Our Linear Regression model achieved an R² score of 0.9014, meaning it explains approximately 90.14% of the variation in sales within the test dataset.
